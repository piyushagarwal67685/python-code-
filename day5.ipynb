```python
#import module
import math

print(math.factorial(5))
print(math.pow(5,2))
print(math.pi)
print(math.floor(10.7234))
print(math.ceil(10.0123))

#example 1
import math as Piyush

print(Piyush.factorial(5))
print(Piyush.pow(5,2))
print(Piyush.pi)
print(Piyush.floor(10.7234))
print(Piyush.ceil(10.0123))

#example 2
from math import factorial,pi

print(factorial(5))
print(pi)

#example 3
from math import *
print(pow(5,3))
print(pi)
print(floor(10.3211))


```
