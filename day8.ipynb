Array reshaping
1. reshape()
2. flatten() -> crete copy
3. ravel() -> return original 
4. trnspose


```python
# reshape array
import numpy as np

arr = np.arange(12)  #12 -> 3*4 || 12 -> 4*3
up_arr = np.reshape(arr, (3, 4))

print(up_arr)
```

```python
#example2d
import numpy as np
arr1 = np.arange(24)

up_arr1 = np.reshape(arr1,(12*2))

print(up_arr1)

#example3d
arr2 = np.arange(24)
up_arr2 = np.reshape(arr2,(3*2*4))
print(up_arr2)
```

```python
#example 2d 
import numpy as np

arr = np.arange(6)
arr_2d = arr.reshape(2, 3)

print(arr_2d)

#example 3d
import numpy as np

arr = np.arange(8)
arr_3d = arr.reshape(2, 2, 2)

print(arr_3d)
```

```python
#flatten 
import numpy as np
arr = np.arange(12).reshape(3, 4)
print(arr)
up_arr = arr.flatten
print(up_arr)

#example 3d
arr1 = np.arange(24).reshape(3, 2, 4)
print(arr1)
up_arr1 = arr1.flatten()
print(up_arr1)
```

```python
#ravel -> works as original array
import numpy as np 
#example 2d
arr = np.arange(14).reshape(7 , 2)
print(arr)
up_arr = arr.ravel()
print(up_arr)


```

```python
#transpose
import numpy as np
arr = np.arange(12).reshape(3, 4)   
print(arr)
up_arr = arr.transpose()
print(up_arr)

#example 3d
import numpy as np

arr = np.arange(8).reshape(2, 2, 2)

print("Original Array:")
print(arr)

up_arr = arr.transpose()

print("\nTranspose Array:")
print(up_arr)
```

numpy array operation 
1. slicing for 1d,2d,3d


```python
#slicing for 1d 
import numpy as np
arr = np.arange(11)
print(arr)
up_arr = arr[:3]
print(up_arr)

#slicing for 2d
import numpy as np
arr1 = np.arange(16).reshape(8,2)
print(arr1)

#slicing for 3d -> arange(24) ->0,23
import numpy as np

arr = np.arange(24).reshape(2,3,4)

print(arr[0,0,0], arr[1,2,3])
```
