```python
def square(x):
  return x * x


#map (2 arguments)
l = [10 , 20 , 30]
l1 = list(map(square,l))
print(l1)


#example
#def square(x):
  #return x * x

  #map (2 arguments)
l = [10 , 20 , 30]
l1 = list(map(lambda x : x * x,l))
print(l1)

#alternative | without in built funtion
l1 = []
for i in range (len(l)):
  l1.append(l[i] * l[i])
print(l1)

#filter
def helo(x):
  return x.startswith('a')

l = ["apple","banana","cat","dog"] #map upper
l1 = list(filter(helo,l))
print(l1)


#example | alternative of filter
l1 = []
for i in l:
  if 'a' == i[-1]:
    l1.append(i)
    print(l1)
```
